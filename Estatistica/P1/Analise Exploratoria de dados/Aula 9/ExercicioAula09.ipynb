# Exercício 1: Análise Exploratória do Banco de Dados `Iris`


**Dataset:** O clássico dataset Iris (flores), disponível publicamente.

**Link:** https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data

**Importação:** Importe as bibliotecas necessárias (pandas, numpy).

*Carregamento dos Dados:* Carregue o dataset Iris a partir da URL fornecida. Defina os nomes das colunas como: ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'Class'].

```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
colnames=['sepal-length', 'sepal-width', 'petal-length', 'petal-width','Class']

iris= pd.read_csv(url, names=colnames)

iris.head(10) # 10 primeiras
```

```python
iris.head(7)
```

```python
iris.info()
```

Use `describe()` para obter um resumo estatístico das colunas numéricas.

```python
# variaveis quantitativas
iris.describe()
```

```python
# variaveis qualitativas
iris.describe(include=['object'])
```

**Seleção e Filtros:**

Selecione apenas as colunas 'sepal-length' e 'petal-length' e exiba as primeiras 5 linhas.

```python
irisSlPl = iris[['sepal-length', 'petal-length']]
irisSlPl.head(5)
```

Filtre e exiba todas as linhas onde o comprimento da sépala (`sepal-length`) é maior que 6.0.

```python
irisMaior60 = iris[iris['sepal-length'] > 6.0]
irisMaior60
```

```python
irisMaior60 = irisMaior60.reset_index()
```

Usando a função `query()`, filtre as linhas onde a largura da pétala (`petal-width`) é menor que 0.5 e a classe é 'Iris-setosa'.

```python
irisPwMenor5eIrisSetosa = iris.query('`petal-width` < 0.5 and Class == "Iris-setosa"')
irisPwMenor5eIrisSetosa
```

```python
irisPwMenor5eIrisSetosa = irisPwMenor5eIrisSetosa.reset_index()
```

**Análise Descritiva:**

Calcule a média, mediana e desvio padrão do comprimento da pétala (`petal-length`) para toda a base de dados.

```python
# média
mediacp = iris['petal-length'].mean()
print(f"A média do comprimento da pétula é: {mediacp:.2f}")
# mediana
medianacp = iris['petal-length'].median()
print(f"A mediana do comprimento da pétula é: {medianacp}")
# desvio padrão
dpcp = iris['petal-length'].std()
print(f"O desvio padrão do comprimento da pétula é: {dpcp:.2f}")
```

Usando `groupby()`, calcule a média do comprimento da sépala (`sepal-length`) para cada classe de flor.

```python
iris_mediasl = iris.groupby('Class')['sepal-length'].mean()
iris_mediasl
```


*Tabelas de Frequência:*

Crie uma tabela de frequência absoluta para a coluna Class.

```python
# Frequência absoluta simples
print("Frequência das classes:")
iris['Class'].value_counts()
```


Crie uma tabela de frequência relativa (percentual) para a coluna Class.

```python
# Frequência relativa simples
print("Frequência relativa das classes:")
iris['Class'].value_counts(normalize=True) * 100
```
