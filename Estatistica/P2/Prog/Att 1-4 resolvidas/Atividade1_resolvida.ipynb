## Questão 1: Troca de variáveis. 
Escreva um programa que receba do usuário duas varáveis a e b, e depois troque os valores das variáveis e imprima os novos valores de a e b. (*Dica: inclua uma terceira variável para ajudar na troca*)

Input

A entrada será:

a -> inteiro

b -> inteiro

Output

A saída deverá ser:

"Valor de a após permutação: {a}"

"Valor de b após permutação: {b}"


Examples

Case: 1

Input
```
a = 45

b = 50
```

Output
```
Valor de a após permutação: 50
Valor de b após permutação: 45
```

````python
A = int(input())
B = int(input())

A, B = B, A

print(f"Valor de a após permutação: {A}")
print(f"Valor de b após permutação: {B}")
````

## Questão 2: 
Escreva um programa que receba do usuário um valor de custo e um valor de venda, mostre o valor do lucro obtido com a venda do produto com duas casa decimais.

Input

A entrada será:

custo -> inteiro

Output

Examples

Case: 1

Input

## Questão 4:

## Questão 5: 
Crie um programa que receba do usuário uma temperatura em Celsius e exiba a temperatura lida usando as escalas Kelvin (K) e Fahrenheit (F).

Considere as seguintes fórmulas:

$$K = C + 273$$

$$F = 1,8\cdot C + 32$$

Input

A entrada será:

celsius -> float


Output

A saída deverá ser:

"Temperatura em Kelvin: {kelvin}"

"Temperatura em Fahrenheit: {fahrenheit}ºF"


Examples

Case: 1

Input
```
celsius = 75.8
```

Output
```
"Temperatura em Kelvin: 348.80K"
"Temperatura em Fahrenheit: 168.44ºF"
```

````python
temp_celsius = float(input())
kelvin = temp_celsius + 273
fahrenheit = (1.8 * temp_celsius) + 32

print(f"Temperatura em Kelvin: {kelvin}\nTemperatura em Fahrenheit: {fahrenheit}ºF")
````

## Questão 6:
Escreva um programa que receba do usuário as 2 notas de um aluno em uma disciplina, depois exiba quantos pontos o aluno ficou distante da nota 10 para cada avaliação, sua média e quantos pontos a média do aluno ficou distante da nota 10.

Input

A entrada será:

nota1 -> float

nota2 -> float


Output

A saída deverá ser:

"A distância da nota 1 para 10 é {dist_n1}"

"A distância da nota 2 para 10 é {dist_n2}"

"A média é {media}"

"A distância da média para 10 é {dist_media}"


Examples

Case: 1

Input
```
Nota1 = 10
Nota2 = 6
```

Output
```
"A distância da nota 1 para 10 é 0.0"
"A distância da nota 2 para 10 é 4.0"
"A média é 8.0"
"A distância da média para 10 é 2.0"
```

## Questão 9:
Em um evento, foram vendidos dois tipos de ingressos:

- Meia-entrada: custa metade do valor do ingresso inteiro.
- Inteira: valor cheio do ingresso.

Sabendo que:

Uma quantidade total de ingressos foi vendida.
Um certo percentual dos ingressos vendidos foi de meia-entrada, e o restante de inteira.
O valor do ingresso inteiro é informado pelo usuário.
Escreva um programa que receba do usuário:

1. A quantidade total de ingressos vendidos;
2. O percentual de ingressos vendidos como meia-entrada (o restante será inteira);
3. O valor do ingresso inteiro.

O programa deve calcular e exibir:

1. A quantidade de ingressos meia-entrada e inteiros vendidos;
2. O valor faturado com cada tipo de ingresso com duas casas decimais;
3. O valor total arrecadado com as vendas com duas casas decimais. 

Input

A entrada será:

ingressos -> inteiro

percentual_meia -> inteiro

valor_ingresso -> float


Output

A saída deverá ser:

"Quantidade de ingressos meia-entrada: {int(total_meia)}"

"Quantidade de ingressos inteiros: {int(total_inteira)}"

"Faturamento com meia-entrada: R${faturamento_meia}"

"Faturamento com inteira: R${faturamento_inteira}"

"Faturamento total: R${faturamento_total}"


Examples

Case: 1

Input
```
ingressos = 500
percentual_meia = 50
valor_ingresso = 120.00
```

Output
```
Quantidade de ingressos meia-entrada: 250
Quantidade de ingressos inteiros: 250
Faturamento com meia-entrada: R$15000.00
Faturamento com inteira: R$30000.00
Faturamento total: R$45000.00
```

````python
ingressos = int(input())
percentual_meia = int(input())
valor_ingresso = float(input())

total_meia = (percentual_meia / 100) * ingressos
total_inteira = ingressos - total_meia
faturamento_meia = total_meia * (valor_ingresso/2)
faturamento_inteira = total_inteira * valor_ingresso
faturamento_total = faturamento_meia + faturamento_inteira

print(f"Quantidade de ingressos meia-entrada: {int(total_meia)}")
print(f"Quantidade de ingressos inteiros: {int(total_inteira)}")
print(f"Faturamento com meia-entrada: R${faturamento_meia:.2f}")
print(f"Faturamento com inteira: R${faturamento_inteira:.2f}")
print(f"Faturamento total: R${faturamento_total:.2f}")
````
