## Questão 1

```
a) (x > y) and (y < 5)        
b) (x == 7) or (y == 7)       
c) not (x < 10)               
d) (x > 10) and (y < 10)     
e) (x > 10) or (y < 10) 
```

## Questão 5

```
fruta = "abacaxi"

print("a" in fruta)
print("x" in fruta)
print("baca" in fruta)
print("uva" in fruta)
print("z" not in fruta)
```
