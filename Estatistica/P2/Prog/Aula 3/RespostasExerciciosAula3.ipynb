## Questão 1
Quais os resultados das expressões abaixo:
```
15 > 10	
8 >= 8	
5 == 7

3 < 1	
20 <= 20
```

## Questão 7
Escreva um programa que leia a idade de uma pessoa e exiba:

"Pode dirigir" se idade >= 18

"Não pode dirigir" caso contrário

````python
idade = int(input("Sua idade: "))
if idade >= 18:
    print("Pode dirigir")
else:
    print("Não pode didigir")   
````
