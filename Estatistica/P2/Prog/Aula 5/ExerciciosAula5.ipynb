# Lista de Exercícios

## Questão 1
Crie uma lista chamada cores com os valores: ["vermelho", "azul", "verde", "amarelo", "roxo"]. Em seguida, exiba:

- O primeiro elemento da lista

- O último elemento da lista (usando índice negativo)

- Os elementos das posições 1 a 3 (índices 1, 2 e 3)

## Questão 2
Dada a lista numeros = [10, 20, 30, 40, 50], realize as seguintes operações:

- Adicione o número 60 no final da lista

- Insira o número 15 na posição 1 (índice 1)

- Remova o número 40 da lista usando `remove()`

- Exiba a lista final

## Questão 3
Crie uma lista com as notas [8, 5, 9, 7, 6, 10] e:

- Exiba a lista ordenada em ordem crescente (sem modificar a original)

- Em seguida, ordene a lista original em ordem decrescente

- Exiba ambas as listas

## Questão 4
Dada a lista letras = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'], utilize fatiamento para:

- Exibir os 3 primeiros elementos

- Exibir os 3 últimos elementos

- Exibir uma cópia da lista inteira

- Exibir os elementos das posições 2 a 5

## Questão 5
Dada a lista valores = [3, 7, 3, 9, 3, 5, 3]`, utilize os métodos adequados para:

- Contar quantas vezes o número 3 aparece na lista

- Encontrar o índice da primeira ocorrência do número 5

- Exibir o maior e o menor valor da lista

- Exibir o tamanho (quantidade de elementos) da lista

## Questão 6
Utilizando list comprehension, crie:

- Uma lista com os números pares de 1 a 20

- Uma lista com o quadrado dos números de 0 a 9

- Uma lista que contenha a palavra "par" para números pares e "ímpar" para números ímpares de 1 a 10

## Questão 7
Crie uma lista `original = [1, 2, 3, 4, 5]` e:

- Faça uma cópia independente usando o método `copy()`

- Faça uma duplicação (atribuição simples) duplicada = original

- Altere o primeiro elemento da lista original para `99`

- Exiba as três listas (original, cópia e duplicada) e explique o resultado

## Questão 8
Crie um programa que:

- Pergunte ao usuário quantas notas ele deseja inserir

- Leia cada uma das notas

- Armazene-as em uma lista

- Calcule e exiba a média das notas

- Exiba quantos alunos tiraram nota acima da média

## Questão 9
Dada a tupla `dados = ("João", 25, 1.75, "São Paulo", 25)`:

- Exiba o nome e a idade da pessoa

- Tente alterar a idade para 26 e explique o que acontece

- Conte quantas vezes o número 25 aparece na tupla

- Encontre o índice da primeira ocorrência de "São Paulo"

- Converta a tupla em uma lista, altere a cidade para "Rio de Janeiro" e converta novamente para tupla

## Questão 10
Crie um programa que:

- Solicite ao usuário que digite seu nome

- Solicite ao usuário que digite sua idade

- Solicite ao usuário que digite sua cidade

- Utilize empacotamento para criar uma tupla chamada pessoa com essas três informações (nesta ordem)

- Utilize desempacotamento para extrair os valores da tupla em variáveis individuais `(nome, idade, cidade)`

- Troque os valores das variáveis `idade` e `cidade` usando desempacotamento de tuplas (sem variável temporária)

- Crie uma nova tupla `pessoa_trocada` com os valores trocados

- Exiba a tupla original e a tupla após a troca
