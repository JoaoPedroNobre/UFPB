## Exercício 9: Teste de mesa
Para o programa abaixo, faça o teste de mesa considerando que o usuário digita 10 para nota1 e 7 para nota2:

```
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2
print("Média:", media)
print("Média arredondada:", round(media, 1))
```

## Exercício 10: Desafio - Conversor de temperaturas
Escreva um programa que:

1. Peça ao usuário uma temperatura em graus Celsius

4. Exiba os resultados formatados com 2 casas decimais

Exemplo de saída:

```
Temperatura em Celsius: 25°C
Em Fahrenheit: 77.00°F
Em Kelvin: 298.15K
```
