## Exercício 5: Conversão de tipos
O código abaixo contém um erro. Identifique e corrija:

```
numero1 = input("Digite o primeiro número: ")
numero2 = input("Digite o segundo número: ")
soma = numero1 + numero2
print("A soma é:", soma)
```
O que acontece quando o usuário digita 5 e 3? Por quê?

**Resposta:** O erro é que `input()` sempre retorna string. As strings "5" e "3" são concatenadas, resultando em "53". Correção:

````python
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))
soma = numero1 + numero2
print("A soma é:", soma)
````

## Exercício 6: Trabalhando com strings
Dada a string `frase = " aprendendo Python é divertido! "`:

1. Exiba o comprimento da string usando len()
2. Remova os espaços das extremidades usando strip()
3. Converta toda a string para maiúsculas
4. Converta toda a string para minúsculas
5. Coloque apenas a primeira letra em maiúscula usando capitalize()
6. Substitua a palavra "divertido" por "incrível"

````python
frase = "  aprendendo Python é divertido!  "

# 1. Exiba o comprimento da string
print(len(frase))  # 34

# 2. Remova os espaços das extremidades
frase_limpa = frase.strip()
print(frase_limpa)  # "aprendendo Python é divertido!"

# 3. Converta para maiúsculas
print(frase_limpa.upper())  # "APRENDENDO PYTHON É DIVERTIDO!"

# 4. Converta para minúsculas
print(frase_limpa.lower())  # "aprendendo python é divertido!"

# 5. Coloque apenas a primeira letra em maiúscula
print(frase_limpa.capitalize())  # "Aprendendo python é divertido!"

# 6. Substitua "divertido" por "incrível"
print(frase_limpa.replace("divertido", "incrível"))  # "aprendendo Python é incrível!"
````

## Exercício 7: F-strings
Crie variáveis para:

```
produto = "Notebook"
preco = 2499.99
quantidade = 3
```

Use f-strings para exibir:

```
"Produto: Notebook"
"Preço unitário: R$ 2499.99"
"Total: R$ 7499.97"
"Compra: 3 x Notebook = R$ 7499.97"
```

````python
produto = "Notebook"
preco = 2499.99
quantidade = 3
total = preco * quantidade

# 1. Produto: Notebook
print(f"Produto: {produto}")

# 2. Preço unitário: R$ 2499.99
print(f"Preço unitário: R$ {preco}")

# 3. Total: R$ 7499.97
print(f"Total: R$ {total}")

# 4. Compra: 3 x Notebook = R$ 7499.97
print(f"Compra: {quantidade} x {produto} = R$ {total:.2f}")
````

## Exercício 8: Métodos de string
Dada a string `email = " USUARIO@EMAIL.COM "`:

1. Remova os espaços das extremidades
2. Converta tudo para minúsculas
3. Verifique se a string começa com "usuario" (use startswith())
4. Verifique se a string contém "@" (use o operador in)
5. Encontre a posição do caractere "@" (use find())
6. Extraia apenas a parte antes do "@" (fatiamento)

````python
email = "   USUARIO@EMAIL.COM   "

# 1. Remova os espaços das extremidades
email_limpo = email.strip()
print(email_limpo)  # "USUARIO@EMAIL.COM"

# 2. Converta tudo para minúsculas
email_minusculo = email_limpo.lower()
print(email_minusculo)  # "usuario@email.com"

# 3. Verifique se começa com "usuario"
print(email_minusculo.startswith("usuario"))  # True

# 4. Verifique se contém "@"
print("@" in email_minusculo)  # True

# 5. Encontre a posição do "@"
posicao_arroba = email_minusculo.find("@")
print(posicao_arroba)  # 7

# 6. Extraia apenas a parte antes do "@"
usuario = email_minusculo[:posicao_arroba]
print(usuario)  # "usuario"
````

## Exercício 10: Desafio - Conversor de temperaturas
Escreva um programa que:

1. Peça ao usuário uma temperatura em graus Celsius
2. Converta para Fahrenheit usando a fórmula: F = (C * 9/5) + 32
3. Converta para Kelvin usando a fórmula: K = C + 273.15
4. Exiba os resultados formatados com 2 casas decimais

Exemplo de saída:

```
Temperatura em Celsius: 25°C
Em Fahrenheit: 77.00°F
Em Kelvin: 298.15K
```

````python
# 1. Peça a temperatura em Celsius
celsius = float(input("Digite a temperatura em graus Celsius: "))

# 2. Converta para Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# 3. Converta para Kelvin
kelvin = celsius + 273.15

# 4. Exiba os resultados formatados
print(f"\nTemperatura em Celsius: {celsius:.2f}°C")
print(f"Em Fahrenheit: {fahrenheit:.2f}°F")
print(f"Em Kelvin: {kelvin:.2f}K")
````
