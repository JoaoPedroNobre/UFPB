# Exercícios da aula 07

## Questão 1
Crie uma função chamada boas_vindas() que exiba a mensagem "Bem-vindo ao meu programa!" quando for chamada. Em seguida, chame a função.


## Questão 2
Crie uma função chamada saudacao(nome) que receba o nome de uma pessoa e exiba a mensagem "Olá, [nome]!". Teste a função com o nome "Maria".

## Questão 3
Crie uma função chamada dobro(numero) que receba um número e retorne o seu dobro. Teste a função com o valor 7 e exiba o resultado.

## Questão 4
Crie uma função chamada `potencia(base, expoente=2)` que calcule a potência de um número. Se o expoente não for informado, eleve ao quadrado. Teste a função com:

- Apenas a base 4 (usando o valor padrão)

- Base 3 e expoente 3

## Questão 5
Crie uma função chamada `calculos(num1, num2)` que retorne a soma, subtração e multiplicação dos dois números. Teste a função com os valores 10 e 3 e exiba os três resultados.



## Questão 6
Use uma função lambda com `map` para calcular o resto da divisão de cada elemento da lista `[17, 23, 42, 8]` por 5. Armazene o resultado em uma lista.

## Questão 7
Dada a lista `cidades = ['João Pessoa', 'Campina Grande', 'Recife', 'Natal']`, utilize `enumerate()` para exibir cada cidade com sua posição no formato: "Posição 0: João Pessoa".

## Questão 8
Dadas as listas:

```
nomes = ['Ana', 'Bruno', 'Carla']
idades = [20, 22, 19]
```

Utilize a função `zip()` para criar um dicionário onde os nomes são as chaves e as idades são os valores. Exiba o dicionário resultante.

## Questão 9
Crie um programa que solicite ao usuário dois números e exiba a divisão do primeiro pelo segundo. Utilize try-except para tratar:

- ValueError: se o usuário digitar algo que não seja número

- ZeroDivisionError: se o usuário tentar dividir por zero

Exiba mensagens de erro amigáveis em cada caso.

## Questão 10
Utilizando a biblioteca `math` (importada de forma adequada), calcule e exiba:

- O valor de $\pi$ (pi)

- A raiz quadrada de 144

- O fatorial de 5

- O valor de 2 elevado à 8ª potência (use math.pow)
