## Questão 4
Crie uma função chamada `potencia(base, expoente=2)` que calcule a potência de um número. Se o expoente não for informado, eleve ao quadrado. Teste a função com:

- Apenas a base 4 (usando o valor padrão)

- Base 3 e expoente 3

````python
def potencia(base, expoente=2):
    return base ** expoente

print(potencia(4))      # 16 (usando valor padrão)
print(potencia(3, 3))   # 27
````

## Questão 5
Crie uma função chamada `calculos(num1, num2)` que retorne a soma, subtração e multiplicação dos dois números. Teste a função com os valores 10 e 3 e exiba os três resultados.



````python
def calculos(num1, num2):
    soma = num1 + num2
    subtracao = num1 - num2
    multiplicacao = num1 * num2
    return soma, subtracao, multiplicacao

s, sub, m = calculos(10, 3)
print(f"Soma: {s}, Subtração: {sub}, Multiplicação: {m}")
# Saída: Soma: 13, Subtração: 7, Multiplicação: 30
````

## Questão 6
Use uma função lambda com `map` para calcular o resto da divisão de cada elemento da lista `[17, 23, 42, 8]` por 5. Armazene o resultado em uma lista.

````python
numeros = [17, 23, 42, 8]
restos = list(map(lambda x: x % 5, numeros))
print(restos)  # [2, 3, 2, 3]
````

## Questão 7
Dada a lista `cidades = ['João Pessoa', 'Campina Grande', 'Recife', 'Natal']`, utilize `enumerate()` para exibir cada cidade com sua posição no formato: "Posição 0: João Pessoa".

````python
cidades = ['João Pessoa', 'Campina Grande', 'Recife', 'Natal']

for indice, cidade in enumerate(cidades):
    print(f"Posição {indice}: {cidade}")
````

## Questão 9
Crie um programa que solicite ao usuário dois números e exiba a divisão do primeiro pelo segundo. Utilize try-except para tratar:

- ValueError: se o usuário digitar algo que não seja número

- ZeroDivisionError: se o usuário tentar dividir por zero

Exiba mensagens de erro amigáveis em cada caso.

````python
try:
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))
    resultado = num1 / num2
    print(f"Resultado: {resultado}")
except ValueError:
    print("Erro: Você deve digitar números válidos!")
except ZeroDivisionError:
    print("Erro: Não é possível dividir por zero!")
````

## Questão 10
Utilizando a biblioteca `math` (importada de forma adequada), calcule e exiba:

- O valor de $\pi$ (pi)

- A raiz quadrada de 144

- O fatorial de 5

- O valor de 2 elevado à 8ª potência (use math.pow)

````python
import math

print(f"Valor de pi: {math.pi}")
print(f"Raiz quadrada de 144: {math.sqrt(144)}")
print(f"Fatorial de 5: {math.factorial(5)}")
print(f"2 elevado à 8ª potência: {math.pow(2, 8)}")
````
