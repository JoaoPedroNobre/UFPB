## Questão 5
Você recebeu os dados de notas de uma turma em três disciplinas. Use o Pandas para calcular as estatísticas descritivas.
```
import pandas as pd
import numpy as np

# Dados das notas
dados = {
    'Aluno': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Eduarda', 'Felipe', 'Gabriela', 'Henrique'],
    'Matemática': [8.5, 6.0, 9.0, 5.5, 7.5, 8.0, 9.5, 6.5],
    'Português': [7.0, 8.5, 6.5, 9.0, 7.0, 6.0, 8.0, 7.5],
    'Ciências': [9.0, 7.5, 8.5, 6.0, 8.0, 7.0, 9.0, 6.5]
}

df = pd.DataFrame(dados)
```
Perguntas:

1. Crie um DataFrame com os dados acima.

2. Calcule a média, mediana e desvio padrão de cada disciplina.

3. Qual disciplina teve a maior média? E a menor?

4. Qual aluno teve a maior nota em Matemática?

5. Calcule a nota média de cada aluno (média das 3 disciplinas) e crie uma nova coluna chamada 'Média Final'.

6. Qual aluno teve a maior média final?

````python
# 1. DataFrame 
import pandas as pd
import numpy as np

# Dados das notas
dados = {
    'Aluno': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Eduarda', 'Felipe', 'Gabriela', 'Henrique'],
    'Matemática': [8.5, 6.0, 9.0, 5.5, 7.5, 8.0, 9.5, 6.5],
    'Português': [7.0, 8.5, 6.5, 9.0, 7.0, 6.0, 8.0, 7.5],
    'Ciências': [9.0, 7.5, 8.5, 6.0, 8.0, 7.0, 9.0, 6.5]
}

df = pd.DataFrame(dados)
print(df)

# 2. Estatísticas
print("As médias são:")
print(df[['Matemática', 'Português', 'Ciências']].mean())
print("As medianas são:")
print(df[['Matemática', 'Português', 'Ciências']].median())
print("Os desvios padrão são:")
print(df[['Matemática', 'Português', 'Ciências']].std())

# 3. Maior média: Ciências (7.875), Menor: Português (7.375)
medias = df[['Matemática', 'Português', 'Ciências']].mean()
print(f"A maior média é: {medias.max()}")

# 4. Aluno com maior nota em Matemática: Gabriela (9.5)
nota_max = df['Matemática'].max()
aluno_max = df[df['Matemática'] == nota_max]['Aluno'].values[0]
print(f"Aluno com maior nota em Matemática é: {aluno_max} ({nota_max:.2f})")

# 5. Média final
df['Média_Final'] = df[['Matemática', 'Português', 'Ciências']].mean(axis=1)
print("As médias finais são:")
print(df['Média_Final'])

# 6. Melhor média final: Gabriela (8.833)
mediafinal_max = df['Média_Final'].max()
posicao_max = df['Média_Final'].idxmax()
aluno_mfmax = df.loc[posicao_max, 'Aluno']
print(f"Aluno com maior média final é: {aluno_mfmax} ({mediafinal_max:.2f})")
````

```
    'Idade': np.random.randint(22, 55, 10),
    'Salario': np.random.randint(3000, 12000, 10)
}

df = pd.DataFrame(dados)
```
Perguntas:

1. Crie o DataFrame com os dados acima.

2. Calcule a média, mediana, mínimo e máximo dos salários.

3. Calcule a média de idade e salário por departamento.

4. Calcule o desvio padrão dos salários por departamento. Qual departamento tem maior variação salarial?

4. Use `df.describe()` para obter um resumo estatístico dos dados numéricos.

5. Crie uma tabela de frequência com a contagem de funcionários por departamento usando `value_counts()`.

````python
# 1. Dataframe
import pandas as pd
import numpy as np

# Dados dos funcionários
np.random.seed(123)
dados = {
    'Nome': ['João', 'Maria', 'Pedro', 'Ana', 'Carlos', 'Juliana', 'Roberto', 'Fernanda', 'Paulo', 'Mariana'],
    'Departamento': np.random.choice(['RH', 'TI', 'Vendas', 'Financeiro'], 10, p=[0.2, 0.3, 0.3, 0.2]),
    'Idade': np.random.randint(22, 55, 10),
    'Salario': np.random.randint(3000, 12000, 10)
}

df = pd.DataFrame(dados)

# 2. Media, mediana, minimo e maxixo dos salarios
print(f"A média do salário é: {df['Salario'].mean()}")
print(f"A mediana do salário é: {df['Salario'].median()}")
print(f"O mínimo do salário é: {df['Salario'].min()}")
print(f"O máximo do salário é: {df['Salario'].max()}")
print("")

# 3. Média por departamento
media_dep = df.groupby('Departamento')[['Idade', 'Salario']].mean()
print(media_dep)
print("")

# 4. Desvio padrão por departamento
dp_dep = df.groupby('Departamento')['Salario'].std()
print(dp_dep)
print("")

# 6. Contagem por departamento
tot_dep = df['Departamento'].value_counts()
print(tot_dep)
````

3. Remova as linhas que têm qualquer valor faltante usando `dropna()`.

4. Substitua os valores faltantes da coluna 'Idade' pela média das idades.

5. Substitua os valores faltantes da coluna 'Renda' pela mediana.

## Questão 9
Uma empresa quer entender melhor o perfil de suas vendas. Faça uma análise descritiva completa.

```
import pandas as pd
import numpy as np

# Dados de vendas por produto
np.random.seed(456)
dados = {
    'Produto': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Preco_Unitario': np.random.uniform(10, 100, 10).round(2),
    'Quantidade_Vendida': np.random.randint(5, 100, 10),
    'Regiao': np.random.choice(['Norte', 'Sul', 'Leste', 'Oeste'], 10),
    'Canal': np.random.choice(['Online', 'Loja', 'Distribuidor'], 10)
}

df = pd.DataFrame(dados)
```

Perguntas:

1. Crie o DataFrame com os dados acima.

2. Use `df.describe()` para obter um resumo estatístico das colunas numéricas.

3. Calcule o valor total de vendas por produto (preço × quantidade) e crie uma nova coluna 'Total_Vendas'.

4. Calcule a média e o desvio padrão de 'Total_Vendas' por região.

5. Calcule a média de 'Preco_Unitario' e 'Quantidade_Vendida' por canal.

6. Identifique o produto com o maior total de vendas usando idxmax().

7. Calcule a mediana, primeiro e terceiro quartis (Q1 e Q3) do 'Preco_Unitario'.

````python
# 1. Dataframe
import pandas as pd
import numpy as np

# Dados de vendas por produto
np.random.seed(456)
dados = {
    'Produto': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Preco_Unitario': np.random.uniform(10, 100, 10).round(2),
    'Quantidade_Vendida': np.random.randint(5, 100, 10),
    'Regiao': np.random.choice(['Norte', 'Sul', 'Leste', 'Oeste'], 10),
    'Canal': np.random.choice(['Online', 'Loja', 'Distribuidor'], 10)
}

df = pd.DataFrame(dados)

# 2. Describe
print(df.describe())
print("")

# 3. Total de vendas
df['Total_Vendas'] = df['Preco_Unitario'] * df['Quantidade_Vendida']
print(df['Total_Vendas'])
print("")

# 4. Média e desvio por região
print(df.groupby('Regiao')['Total_Vendas'].agg(['mean', 'std']))
print("")

# 5. Média por canal
print(df.groupby('Canal')[['Preco_Unitario', 'Quantidade_Vendida']].mean())
print("")

# 6. Produto com maior venda
print(df.loc[df['Total_Vendas'].idxmax(), 'Produto'])
print("")

# 7. Quartis do Preço
print(df['Preco_Unitario'].quantile([0, 0.25, 0.5, 0.75, 1]))
````
