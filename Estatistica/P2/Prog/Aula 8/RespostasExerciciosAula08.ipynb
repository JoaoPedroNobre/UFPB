## Questão 4
Dado o array `arr = np.array([[1, 2, 3], [4, 5, 6]])`, responda:

- Quantas dimensões (`.ndim`) ele possui?

- Qual é o seu formato (`.shape`)?

- Qual é o tipo dos seus elementos (`.dtype`)?

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(f".ndim: {arr.ndim}")
print(f".shape: {arr.shape}")
print(f".dtype: {arr.dtype}")
```

## Questão 5
Crie um array `arr = np.array([2, 4, 6, 8, 10])` e realize as seguintes operações sem usar loops:

- Multiplique todos os elementos por 3

- Adicione 10 a cada elemento

- Calcule o quadrado de cada elemento

```python
arr = np.array([2, 4, 6, 8, 10])
print(arr * 3)
print(arr + 10)
print(arr ** 2)
```

## Questão 6
Dado o array `arr = np.array([100, 200, 300, 400, 500, 600])`:

- Acesse o primeiro elemento (índice 0)

- Acesse o último elemento usando índice negativo

- Acesse os elementos das posições 2 a 4 (índices 2, 3 e 4).



```python
arr = np.array([100, 200, 300, 400, 500, 600])
print(arr[0])      # 100
print(arr[-1])     # 600
print(arr[2:5])    # [300 400 500]
```

## Questão 7
Crie um array `original = np.array([1, 2, 3, 4, 5])`. Em seguida:

- Crie uma view (fatia) com os 3 primeiros elementos

- Modifique o primeiro elemento da view para 99

- Exiba o array original e explique o que aconteceu.

```python
original = np.array([1, 2, 3, 4, 5])
view = original[:3]
view[0] = 99
print(original)  # [99  2  3  4  5] - o original também mudou porque view é uma referência, não uma cópia
```
