## Questão 1
Escreva um programa que receba dois números de um usuário, esses números serão usados em uma função que receba dois parâmetros: número e potência. O retorno dessa função deve ser o numero elevado à potência

Input

As entradas deverão ser:

numero -> float
potencia -> inteiro
Output

A saída deverá ser o número elevado à potência, com duas casas decimais.

Examples

```
Case: 1

Input

2
2

Output

4.00

Case: 2

Input

18.76
4

Output

123860.07
```

## Questão 2
Escreva um programa que receba três valores de um usuário. Esses valores serão usados em uma função chamada media(a, b, c) que recebe três números como argumentos e retorna a média aritmética entre eles.

Input

As entradas serão:

n1 -> float
n2 -> float
n3 -> float
Output

A saída deverá ser a média aritmética entre os três números, com duas casas decimais.

Examples

```
Case: 1

Input

9
12
4

Output

8.33

Case: 2

Input

8
8
8

Output

8.00

Case: 3

Input

2
4
6

Output

4.00

Case: 4

Input

17.5874
14.21
13.5877

Output

15.13
```

## Questão 3
Escreva um programa que receba dois valores de um usuário. Esses valores serão usados em uma função `area_retangulo(base, altura)` que calcule e retorne a área de um retângulo.

Input

As entradas serão:

base -> float

altura -> float

Output

A saída deverá ser:

"A área do retângulo é {area}".

Obs.: area deve ter duas casas decimais

Examples

```
Case: 1

Input

40
21

Output

A área do retângulo é 840.00

Case: 2

Input

10
10

Output

A área do retângulo é 100.00

Case: 3

Input

12.47
5.2874

Output

A área do retângulo é 65.93
```

## Questão 4
Faça um programa que, dado uma string fornecida por um usuário, conte a quantidade de vogais que aparecem naquela palavra.

Obs.: letras maiusculas e minusculas devem ser consideradas, mas letras acentuadas não.

Input

A entrada será:

word -> string
Output

A saída deverá ser:

"A quantidade de vogais em {palavra} é {resultado_contagem}"

Examples

```
Case: 1

Input

Macaco Elétrico

Output

A quantidade de vogais em Macaco Elétrico é 6

Case: 2

Input

Programe.py é muito legal, programação é massa!

Output

A quantidade de vogais em Programe.py é muito legal, programação é massa! é 14
```

## Questão 5
Crie uma função `alunos_acima_da_media(nomes, notas)` que recebe duas listas com os nomes e as notas dos alunos fornecidas por um usuário. Após isso, o programa deverá ser capaz de exibir o nome dos alunos que foram aprovados (nota >= 7)

Input

As entradas serão:

n -> inteiro

E depois, de 0 a n-1, receber:

nome_do_aluno -> string

nota_do_aluno -> float

Output

A saída deverá ser:

"Os alunos que foram aprovados foram: "

Seguido do nome dos alunos que foram aprovados, linha após linha.

Examples

```
Case: 1

Input

2
Ana
4
Josué
8

Output

Os alunos que foram aprovados foram:
Josué

Case: 2

Input

4
Ana
7
Karla
5
João
8.5
Marcondes
7.0

Output

Os alunos que foram aprovados foram:
Ana
João
Marcondes

Case: 3

Input

5
João
10
Urias
9
Anitta
1
Alice
8
Marcelo
10

Output

Os alunos que foram aprovados foram:
João
Urias
Alice
Marcelo
```

## Questão 6
Faça um programa que, dado uma string fornecida por um usuário, crie uma função que conta quantas vezes cada vogal aparece na string.

Obs.: letras maiusculas e minusculas são consideradas, mas letras acentuadas não são.

Input

A entrada será:

word -> string

Output

A saída deverá ser um dicionário, onde a chave corresponde a vogal e o valor é a quantidade de vezes que aquela vogal apareceu na variável word.

Examples

```
Case: 1

Input

Macaco Elétrico

Output

{'a': 2, 'e': 1, 'i': 1, 'o': 2, 'u': 0}

Case: 2

Input

Uva e Maçã e Pinha

Output

{'a': 3, 'e': 2, 'i': 1, 'o': 0, 'u': 1}
```
