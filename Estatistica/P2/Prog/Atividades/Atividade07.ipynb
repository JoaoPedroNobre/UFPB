````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
np.random.seed(42)

# dados
dados_vendas = {
    'Loja': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Curitiba',
               'Porto Alegre', 'São Paulo', 'Rio de Janeiro', 'Belo Horizonte',
               'Curitiba', 'Porto Alegre'],
    'Categoria': ['Eletrônicos', 'Roupas', 'Alimentos', 'Eletrônicos', 'Roupas',
                  'Alimentos', 'Eletrônicos', 'Roupas', 'Alimentos', 'Eletrônicos'],
    'Vendas_Jan': np.random.randint(100, 500, 10),
    'Vendas_Fev': np.random.randint(100, 500, 10),
    'Vendas_Mar': np.random.randint(100, 500, 10),
    'Clientes': np.random.randint(50, 200, 10),
    'Ticket_Medio': np.random.uniform(50, 300, 10).round(2)
}

df_vendas = pd.DataFrame(dados_vendas)
df_vendas['Total_Vendas'] = df_vendas[['Vendas_Jan', 'Vendas_Fev', 'Vendas_Mar']].sum(axis=1)

# analise
print("--- Estrutura do DataFrame ---")
df_vendas.info()
print("\n--- Primeiras Linhas ---")
print(df_vendas.head())
print("\n--- Estatísticas Descritivas ---")
print(df_vendas.describe())

# novas colunas
df_vendas['media_mensal'] = (df_vendas['Total_Vendas'] / 3).round(2)
df_vendas['Receita_Total'] = df_vendas['Total_Vendas'] * df_vendas['Ticket_Medio']

# filtro de lojas com vendas acima de 200 em Janeiro
df_filtrado = df_vendas[df_vendas['Vendas_Jan'] > 200]

# venda por loja cidade categoria
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# total de vendas de cada loja
sns.barplot(data=df_vendas, x='Loja', y='Total_Vendas', palette='viridis', ax=axes[0], hue='Loja', legend=False)
for p in axes[0].patches:
    axes[0].annotate(f'{p.get_height():.0f}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha='center', va='bottom')
axes[0].set_title('Total de Vendas por Loja')

# comparação por cidade (media)
cidade_media = df_vendas.groupby('Cidade', as_index=False)['Total_Vendas'].mean().sort_values(by='Total_Vendas')
sns.barplot(data=cidade_media, y='Cidade', x='Total_Vendas', color='skyblue', ax=axes[1])
axes[1].set_title('Média de Total de Vendas por Cidade')

# comparação por categoria (media do ticket medio)
sns.barplot(data=df_vendas, x='Categoria', y='Ticket_Medio', palette='pastel', errorbar=None, ax=axes[2], hue='Categoria', legend=False)
axes[2].set_title('Ticket Médio por Categoria')

plt.tight_layout()
plt.show()

# distri vendas em pizza
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 1. Participação por cidade (Explode no maior valor)
vendas_cidade = df_vendas.groupby('Cidade')['Total_Vendas'].sum()
explode = [0.1 if v == vendas_cidade.max() else 0 for v in vendas_cidade]
axes[0].pie(vendas_cidade, labels=vendas_cidade.index, autopct='%1.1f%%', explode=explode, startangle=90, colors=sns.color_palette('Set2'))
axes[0].set_title('Participação de Vendas por Cidade')

# 2. Receita Total por Categoria
receita_cat = df_vendas.groupby('Categoria')['Receita_Total'].sum()
axes[1].pie(receita_cat, labels=receita_cat.index, autopct='%1.1f%%', colors=sns.color_palette('pastel'), startangle=90)
axes[1].set_title('Receita Total por Categoria')

plt.show()

# relacao entre clientes e vendas
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# regressão
sns.regplot(data=df_vendas, x='Clientes', y='Total_Vendas', ax=axes[0], color='indigo')
corr_cli_vendas = df_vendas['Clientes'].corr(df_vendas['Total_Vendas'])
axes[0].set_title(f'Regressão (Correlação: {corr_cli_vendas:.2f})')

# scatter 
sns.scatterplot(data=df_vendas, x='Clientes', y='Total_Vendas', size='Ticket_Medio', 
                hue='Categoria', sizes=(50, 300), alpha=0.7, ax=axes[1])
axes[1].set_title('Clientes vs Total Vendas (Bolhas = Ticket Médio)')
axes[1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

# pairplot e heatmap
cols_num = ['Clientes', 'Total_Vendas', 'Ticket_Medio', 'Vendas_Jan', 'Vendas_Fev', 'Vendas_Mar']

# Pairplot (Cria uma figura independente)
sns.pairplot(df_vendas[cols_num + ['Cidade']], hue='Cidade', corner=True)
plt.show()

# Heatmap
plt.figure(figsize=(8, 6))
matriz_corr = df_vendas[cols_num].corr()
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
plt.title('Matriz de Correlação')
plt.show()

# evolucao mensal
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# evolucao por cidade
df_melt = df_vendas.melt(id_vars=['Cidade'], value_vars=['Vendas_Jan', 'Vendas_Fev', 'Vendas_Mar'], 
                         var_name='Mes', value_name='Vendas')
sns.lineplot(data=df_melt, x='Mes', y='Vendas', hue='Cidade', marker='o', errorbar=None, ax=axes[0])
axes[0].set_title('Evolução das Vendas Mensais por Cidade')

# tendencia media geral
media_mensal = df_vendas[['Vendas_Jan', 'Vendas_Fev', 'Vendas_Mar']].mean()
axes[1].plot(media_mensal.index, media_mensal.values, marker='s', color='firebrick', linewidth=2)
axes[1].set_title('Média de Vendas Mensais (Geral)')

# Anotação do pico
mes_max = media_mensal.idxmax()
val_max = media_mensal.max()
axes[1].annotate('Maior Venda', xy=(mes_max, val_max), xytext=(mes_max, val_max + 20),
                 arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
                 ha='center')

plt.show()

# histograma e box plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histograma
sns.histplot(df_vendas['Total_Vendas'], bins=10, kde=True, color='teal', ax=axes[0])
mediana_vendas = df_vendas['Total_Vendas'].median()
axes[0].axvline(mediana_vendas, color='red', linestyle='--', label=f'Mediana ({mediana_vendas:.1f})')
axes[0].legend()
axes[0].set_title('Distribuição do Total de Vendas')

# Boxplot
sns.boxplot(data=df_vendas, x='Categoria', y='Total_Vendas', palette='Set3', ax=axes[1], hue='Categoria', legend=False)
axes[1].set_title('Distribuição de Vendas por Categoria')

plt.show()

# categorica
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df_vendas, x='Cidade', palette='Blues_r', ax=axes[0], hue='Cidade', legend=False)
axes[0].set_title('Quantidade de Lojas por Cidade')
axes[0].tick_params(axis='x', rotation=45)

sns.countplot(data=df_vendas, x='Cidade', hue='Categoria', palette='pastel', ax=axes[1])
axes[1].set_title('Lojas por Cidade e Categoria')
axes[1].tick_params(axis='x', rotation=45)
axes[1].legend(title='Categoria')

plt.tight_layout()
plt.show()
````
