## Questão 6
Escreva um programa para recerber de um usuário o tamanho de três segmentos de retas e informar se estas podem formar um triângulo (a soma de quaisquer dois lados sempre deve ser maior que o terceiro lado). Caso sim, também deve-se informar se este é equilátero (tem os três lados iguais), escaleno (tem os três lados diferentes) ou isósceles (tem apenas dois lados iguais).

Input

A entrada será:

lado_1 -> float

lado_2 -> float

lado_3 -> float



Output

A saída deverá ser:

"Esses segmentos de reta não formam um triângulo. Tchau!!!!"

ou 

"Triangulinho com todos os lados iguais: é equilátero!!"

ou

"Triangulinho com dois lados iguais: é isósceles!!"

ou

"Tem os três lados diferentes mas ainda continua sendo um triangulinho: é escaleno!!"



Examples

Case: 1

Input
```
lado_1 = 15
lado_2 = 28
lado_3 = 10
```

Output 
```
"Esses segmentos de reta não formam um triângulo. Tchau!!!!"
```

Case: 2

Input
```
lado_1 = 6
lado_2 = 6
lado_3 = 4
```

Output 
```
"Triangulinho com dois lados iguais: é isósceles!!"
```

Case: 3

Input
```
lado_1 = 6
lado_2 = 6
lado_3 = 6
```

Output 
```
"Triangulinho com todos os lados iguais: é equilátero!!"
```

Case: 4

Input
```
lado_1 = 6
lado_2 = 5
lado_3 = 4
```

Output 
```
"Tem os três lados diferentes mas ainda continua sendo um triangulinho: é escaleno!!"
```

````python
lado_1 = float(input())
lado_2 = float(input())
lado_3 = float(input())

forma_triangulo = (
    lado_1 > 0 and lado_2 > 0 and lado_3 > 0
    and lado_1 + lado_2 > lado_3
    and lado_1 + lado_3 > lado_2
    and lado_2 + lado_3 > lado_1
)
if not forma_triangulo:
    print("Esses segmentos de reta não formam um triângulo. Tchau!!!!")
elif lado_1 == lado_2 == lado_3:
    print("Triangulinho com todos os lados iguais: é equilátero!!")
elif lado_1 == lado_2 or lado_1 == lado_3 or lado_2 == lado_3:
    print("Triangulinho com dois lados iguais: é isósceles!!")
else:
    print("Tem os três lados diferentes mas ainda continua sendo um triangulinho: é escaleno!!")
````

Faça um programa que, dado um valor n recebido por um usuário, exiba os valores de 0 a n que são múltiplos de 3, de 5 e de ambos.

Input

A entrada será:

n -> inteiro

Output

0 é múltiplo de 3 e de 5

3 é múltiplo de 3

5 é múltiplo de 5

6 é múltiplo de 3

9 é múltiplo de 3

10 é múltiplo de 5
....

Examples


Case: 2

## Questão 10

Faça um programa para receber de um usuário números inteiros e positivos até o usuário não desejar continuar informando um novo número. Ao final, o programa deve exibir o total de números lidos e a soma deles.

Input

A entrada deverá ser:

confirmacao -> string

numero -> float

Obs.: A mensagem de confirmação é um "S" ou "s". Caso a variável confirmacao seja qualquer outra coisa, encerrar o código.

Output

A saída deverá ser:

"Você digitou {quantidade} números"

"A soma final é: {soma}"

Obs.: soma deve ter 2 casas decimais

Examples

Case: 1

Input
```
s
10
s
20
n
```

Output
```
Você digitou 2 números
A soma final é: 30.00
```

Case: 2

Input
```
n
```

Output
```
Você digitou 0 números
A soma final é: 0.00
```

Case: 3

Input
```
s
50
s
20
s
10
s
10
n
```

Output
```
Você digitou 4 números
A soma final é: 90.00
```

````python
quantidade = 0
soma = 0.0
confirmacao = input()

while confirmacao.lower() == "s":
    numero = float(input())
    quantidade += 1
    soma += numero
    confirmacao = input()

print(f"Você digitou {quantidade} números")
print(f"A soma final é: {soma:.2f}")
````

## Questão 11
A sequência de Fibonacci é uma famosa sequência numérica onde cada número é a soma dos dois anteriores. Ela começa com os termos 0 e 1, e segue assim:

0, 1, 1, 2, 3, 5, 8, 13, 21, ...

Ou seja:

O 0º termo é 0

O 1º termo é 1

A partir do 2º termo: F(n) = F(n-1) + F(n-2)

Faça um programa que, dado um número inteiro n, imprima todos os termos da sequência de Fibonacci menores ou iguais a n, linha a linha.

Examples

Case: 1

```
Input

```

```

```


## Questão 12
Faça um programa que, dado um saldo inicial recebdo de um usuário, simule operações de depositar ou sacar.

Input

A entrada será a seguinte:

saldo -> float

operacao -> string

Enquanto o usuário não colocar como entrada "sair" na *operacao*, o programa deverá ser capaz de executar as operação de depositar ou sacar.

Obs.: a operação de depositar só será feita se a variável operacao for "depositar" e a operação de sacar só será feita se a variável operacao for sacar.

Após verificar se a operação é válida ou não, seu programa deve receber o valor (float) que será depositado/sacado.

Output

Ao final, seu programa deverá mostrar:

"Seu saldo era: R$ {saldo_anterior}"

"Seu novo saldo é: R$ {saldo_atual}"

As variáveis devem ter 2 casas decimais.

Examples

Case: 1

```
Input

1500
depositar
20
sacar
20
sair

Output

Seu saldo era: R$ 1500.00
Seu novo saldo é: R$ 1500.00
```

Case: 2

```
Input

250.21
sacar
50.21
depositar
200
depositar
500
sair

Output

Seu saldo era: R$ 250.21
Seu novo saldo é: R$ 900.00
```

Case: 3

```
Input

3600.48
sacar
3600.48
sacar
2
sair

Output

Seu saldo era: R$ 3600.48
Seu novo saldo é: R$ -2.00
```

````python
saldo = float(input())
saldo_anterior = saldo

while True:
    operacao = input()
    if operacao == "sair":
        break
    if operacao == "depositar":
        valor = float(input())
        saldo += valor
    elif operacao == "sacar":
        valor = float(input())
        saldo -= valor
 
print(f"Seu saldo era: R$ {saldo_anterior:.2f}")
print(f"Seu novo saldo é: R$ {saldo:.2f}")
````
