## Questão 1
Você recebeu os dados das notas de 10 alunos em 3 disciplinas diferentes. Os dados estão organizados em uma lista de listas (cada sublista representa um aluno com [Nome, Matemática, Física, Química]).
```
dados_alunos = [
    ["Ana", 7.5, 8.0, 6.5],
    ["João", 5.0, 6.5, 7.0],
    ["Maria", 9.0, 8.5, 9.5],
    ["Pedro", 6.0, 5.5, 6.0],
    ["Carla", 8.5, 9.0, 8.0],
    ["Lucas", 4.5, 5.0, 6.5],
    ["Julia", 9.5, 9.0, 8.5],
    ["Rafael", 6.5, 7.0, 7.5],
    ["Beatriz", 8.0, 8.5, 9.0],
    ["Thiago", 5.5, 6.0, 5.0]
]
```
Utilizando NumPy, faça:

a) Criação do array: Converta os dados numéricos (apenas as notas) em um array NumPy 2D de tamanho 10x3.

b) Estatísticas por aluno: Calcule a média, o desvio padrão e a nota máxima de cada aluno. Armazene os resultados em três arrays diferentes.

c) Classificação dos alunos: Crie uma função que receba a média de um aluno e retorne:

- "Aprovado" se média ≥ 7.0

- "Recuperação" se 5.0 ≤ média < 7.0

- "Reprovado" se média < 5.0

Aplique esta função a todos os alunos usando uma estrutura de repetição.

d) Relatório final (0,5 ponto): Gere um relatório que exiba:
```
Nome     | Matemática | Física | Química | Média | Situação
-------------------------------------------------------------
Ana      |    7.5     |  8.0   |   6.5   |  7.33 | Aprovado
```
Além disso, exiba a média geral da turma e o aluno com maior média.

````python
import numpy as np


dados_alunos = [
    ["Ana", 7.5, 8.0, 6.5],
    ["João", 5.0, 6.5, 7.0],
    ["Maria", 9.0, 8.5, 9.5],
    ["Pedro", 6.0, 5.5, 6.0],
    ["Carla", 8.5, 9.0, 8.0],
    ["Lucas", 4.5, 5.0, 6.5],
    ["Julia", 9.5, 9.0, 8.5],
    ["Rafael", 6.5, 7.0, 7.5],
    ["Beatriz", 8.0, 8.5, 9.0],
    ["Thiago", 5.5, 6.0, 5.0],
]


notas = np.array([linha[1:] for linha in dados_alunos], dtype=float)

medias = np.mean(notas, axis=1)
desvios = np.std(notas, axis=1)
maximos = np.max(notas, axis=1)


def classificar_aluno(media: float) -> str:
    if media >= 7.0:
        return "Aprovado"
    elif media >= 5.0:
        return "Recuperação"
    else:
        return "Reprovado"


situacoes = [classificar_aluno(m) for m in medias]

print(
    f"{'Nome':<10} | {'Matemática':<10} | {'Física':<8} | {'Química':<8} | {'Média':<6} | {'Situação':<11}"
)
print("-" * 68)

for i, linha in enumerate(dados_alunos):
    nome = linha[0]
    mat, fis, qui = notas[i]
    media_aluno = medias[i]
    situacao = situacoes[i]
    print(
        f"{nome:<10} | {mat:<10.1f} | {fis:<8.1f} | {qui:<8.1f} | {media_aluno:<6.2f} | {situacao:<11}"
    )

print("-" * 68)

media_geral = np.mean(notas)
idx_maior_media = np.argmax(medias)
nome_maior_media = dados_alunos[idx_maior_media][0]
maior_media = medias[idx_maior_media]

print(f"\nMédia geral da turma: {media_geral:.2f}")
print(f"Aluno com a maior média: {nome_maior_media} ({maior_media:.2f})")
````

## Questão 2
Uma loja aplica descontos diferentes para produtos de categorias distintas. Os preços dos produtos estão em um array 2D (linhas = produtos, colunas = meses). Os percentuais de desconto variam por categoria.
```
# Array de preços: 5 produtos x 4 meses
precos = np.array([
    [100, 110, 105, 120],   # Produto A
    [200, 190, 195, 210],   # Produto B
    [50, 55, 52, 58],       # Produto C
    [300, 310, 305, 320],   # Produto D
    [150, 145, 155, 160]    # Produto E
])

# Percentuais de desconto por categoria (0 = sem desconto)
desconto_categoria = np.array([0.10, 0.05, 0.00, 0.15, 0.08])  # 10%, 5%, 0%, 15%, 8%
```
Responda:

a) Aplicação do desconto: Utilizando broadcasting, crie um novo array `precos_com_desconto` onde cada produto (linha) tenha seu preço reduzido pelo percentual correspondente à sua categoria. Use apenas operações vetorizadas (sem loops).

b) Análise mensal: Para cada mês (coluna), calcule:

- O preço médio após desconto

- O produto mais caro (índice da linha)

- O produto mais barato (índice da linha)

c) Filtro de produtos: Crie uma máscara booleana para identificar quais produtos, em algum mês, tiveram preço final superior a 250. Exiba esses produtos e seus respectivos preços.

d) Função de análise: Crie uma função chamada `analisar_produto' que receba o array de preços originais e o índice de um produto. A função deve retornar um dicionário com: preço médio original, preço médio com desconto, maior preço original, menor preço com desconto. Teste a função para o produto D (índice 3).

````python
import numpy as np


precos = np.array(
    [
        [100, 110, 105, 120],  
        [200, 190, 195, 210],  
        [50, 55, 52, 58],  
        [300, 310, 305, 320],  
        [150, 145, 155, 160],  
    ]
)

desconto_categoria = np.array([0.10, 0.05, 0.00, 0.15, 0.08])
nomes_produtos = ["Produto A", "Produto B", "Produto C", "Produto D", "Produto E"]


precos_com_desconto = precos * (1 - desconto_categoria[:, np.newaxis])


media_mensal = np.mean(precos_com_desconto, axis=0)
produto_mais_caro_idx = np.argmax(precos_com_desconto, axis=0)
produto_mais_barato_idx = np.argmin(precos_com_desconto, axis=0)



mascara_250 = np.any(precos_com_desconto > 250, axis=1)
produtos_filtrados = precos_com_desconto[mascara_250]


def analisar_produto(
    precos_originais: np.ndarray,
    indice: int,
    descontos: np.ndarray = desconto_categoria,
) -> dict:
    preco_orig = precos_originais[indice]
    preco_desc = preco_orig * (1 - descontos[indice])

    return {
        "preço médio original": float(np.mean(preco_orig)),
        "preço médio com desconto": float(np.mean(preco_desc)),
        "maior preço original": float(np.max(preco_orig)),
        "menor preço com desconto": float(np.min(preco_desc)),
    }



resultado_produto_d = analisar_produto(precos, 3)


print("--- a) Preços com Desconto ---")
print(np.round(precos_com_desconto, 2))
print()

print("--- b) Análise Mensal ---")
for mes in range(4):
    print(f"Mês {mes + 1}:")
    print(f"  Preço médio após desconto: R$ {media_mensal[mes]:.2f}")
    print(
        f"  Mais caro (índice {produto_mais_caro_idx[mes]}): {nomes_produtos[produto_mais_caro_idx[mes]]}"
    )
    print(
        f"  Mais barato (índice {produto_mais_barato_idx[mes]}): {nomes_produtos[produto_mais_barato_idx[mes]]}"
    )
print()

print("--- c) Produtos com preço final > 250 em algum mês ---")
indices_filtrados = np.where(mascara_250)[0]
for idx in indices_filtrados:
    print(
        f"{nomes_produtos[idx]} (Índice {idx}): {precos_com_desconto[idx].round(2)}"
    )
print()

print("--- d) Análise do Produto D (Índice 3) ---")
for chave, valor in resultado_produto_d.items():
    print(f"{chave}: {valor:.2f}")
````

```
vendas_outubro = {

```
Responda:

a) Criação da Series: Converta o dicionário em uma Series do Pandas com nome "Vendas Outubro 2024". Utilize os dias como índice.

b) Estatísticas com métodos: Utilizando métodos do Pandas, calcule e exiba:

- Total de vendas do mês

- Média de vendas diárias

- Mediana das vendas

c) Filtros condicionais: Utilize indexação booleana para responder:

- Quantos dias tiveram vendas acima de 2000?

- Qual a média de vendas nos dias com venda superior a 2000?

- Quais dias tiveram vendas entre 1500 e 2000? (use operadores `&` ou `|`)

d) Função de classificação: Crie uma função classificar_dia que receba o valor da venda e retorne:

- "Excelente" se venda > 2500

- "Bom" se 2000 ≤ venda ≤ 2500

- "Regular" se 1000 ≤ venda < 2000

- "Ruim" se venda < 1000

Aplique esta função à Series usando o método `.apply()` e crie uma nova Series com a classificação de cada dia.

````python
import pandas as pd


vendas_outubro = {
    "2024-10-01": 1250,
    "2024-10-02": 1340,
    "2024-10-03": 980,
    "2024-10-04": 2100,
    "2024-10-05": 1870,
    "2024-10-08": 1560,
    "2024-10-09": 1430,
    "2024-10-10": 1890,
    "2024-10-11": 1780,
    "2024-10-12": 3400,
    "2024-10-15": 1670,
    "2024-10-16": 1550,
    "2024-10-17": 2290,
    "2024-10-18": 1980,
    "2024-10-19": 2750,
    "2024-10-22": 1430,
    "2024-10-23": 1870,
    "2024-10-24": 2340,
    "2024-10-25": 2210,
    "2024-10-26": 1980,
    "2024-10-29": 1670,
    "2024-10-30": 2450,
    "2024-10-31": 3100,
}


vendas = pd.Series(vendas_outubro, name="Vendas Outubro 2024")


total_vendas = vendas.sum()
media_vendas = vendas.mean()
mediana_vendas = vendas.median()
desvio_padrao = vendas.std()
dia_maior_venda = vendas.idxmax()
dia_menor_venda = vendas.idxmin()



acima_2000 = vendas[vendas > 2000]
qtd_dias_acima_2000 = len(acima_2000)
media_acima_2000 = acima_2000.mean()


dias_entre_1500_2000 = vendas[(vendas >= 1500) & (vendas <= 2000)]


def classificar_dia(venda: float) -> str:
    if venda > 2500:
        return "Excelente"
    elif 2000 <= venda <= 2500:
        return "Bom"
    elif 1000 <= venda < 2000:
        return "Regular"
    else:
        return "Ruim"


classificacao_dias = vendas.apply(classificar_dia)


print("=== a) Series Criada ===")
print(vendas.head(3))
print("...\n")

print("=== b) Estatísticas ===")
print(f"Total de vendas no mês: R$ {total_vendas:,.2f}")
print(f"Média de vendas diárias: R$ {media_vendas:,.2f}")
print(f"Mediana das vendas: R$ {mediana_vendas:,.2f}")
print(f"Desvio padrão das vendas: R$ {desvio_padrao:,.2f}")
print(
    f"Dia com maior venda: {dia_maior_venda} (R$ {vendas[dia_maior_venda]:,.2f})"
)
print(
    f"Dia com menor venda: {dia_menor_venda} (R$ {vendas[dia_menor_venda]:,.2f})\n"
)

print("=== c) Filtros Condicionais ===")
print(f"Quantidade de dias com vendas > 2000: {qtd_dias_acima_2000}")
print(f"Média das vendas nos dias > 2000: R$ {media_acima_2000:,.2f}\n")
print("Dias com vendas entre 1500 e 2000:")
print(dias_entre_1500_2000)
print()

print("=== d) Classificação dos Dias ===")
print(classificacao_dias)
````

## Questão 4
Uma escola tem duas turmas (A e B) que fizeram as mesmas provas. As notas estão armazenadas em Series com índices dos alunos.
```
import pandas as pd

# Turma A - 15 alunos
turma_a = pd.Series([8.5, 7.0, 9.5, 6.0, 8.0, 7.5, 9.0, 5.5, 8.5, 7.0, 9.5, 6.5, 8.0, 7.0, 9.0],
                    index=['Ana', 'João', 'Maria', 'Pedro', 'Carla', 'Lucas', 'Julia', 
                           'Rafael', 'Beatriz', 'Thiago', 'Amanda', 'Bruno', 'Fernanda', 
                           'Gustavo', 'Patrícia'])

# Turma B - 12 alunos (alguns nomes diferentes, outros coincidentes)
turma_b = pd.Series([9.0, 6.5, 8.5, 7.5, 9.5, 6.0, 8.0, 7.0, 9.0, 6.5, 8.5, 7.5],
                    index=['Ana', 'Pedro', 'Carla', 'Lucas', 'Julia', 'Rafael', 
                           'Beatriz', 'Thiago', 'Fernanda', 'Roberto', 'Camila', 'Daniel'])
```
Responda:

a) Criação e inspeção: Crie as duas Series e identifique:

- Quantos alunos existem apenas na Turma A (não estão na B)?

- Quantos alunos existem apenas na Turma B (não estão na A)?

- Quantos alunos estão em ambas as turmas?

**Dica:** Use os métodos `.index` e operações de conjunto ou `.isin()`

b) Comparação de notas: Calcule a diferença de notas entre as duas turmas para os alunos em comum. A Series resultante deve conter apenas os alunos que estão nas duas turmas, com a diferença (Nota_TurmaA - Nota_TurmaB). Utilize o alinhamento automático do Pandas.

c) Estatísticas comparativas: Calcule e responda:

- Qual a média da Turma A? E da Turma B?

- Considerando apenas os alunos em comum, qual turma teve melhor desempenho médio?

- Qual aluno teve a maior diferença positiva (Turma A muito melhor que B)?

- Qual aluno teve a maior diferença negativa (Turma B muito melhor que A)?

d) Tratamento de valores faltantes:

- Some as duas Series (como se fosse uma nota total se o aluno fez ambas as provas). O que acontece com os alunos que estão em apenas uma turma?

- Utilizando o método `.add()` com o argumento `fill_value=0`, calcule a soma tratando valores ausentes como zero.

Explique a diferença entre os dois resultados.

````python
import pandas as pd

# Dados fornecidos
turma_a = pd.Series(
    [
        8.5,
        7.0,
        9.5,
        6.0,
        8.0,
        7.5,
        9.0,
        5.5,
        8.5,
        7.0,
        9.5,
        6.5,
        8.0,
        7.0,
        9.0,
    ],
    index=[
        "Ana",
        "João",
        "Maria",
        "Pedro",
        "Carla",
        "Lucas",
        "Julia",
        "Rafael",
        "Beatriz",
        "Thiago",
        "Amanda",
        "Bruno",
        "Fernanda",
        "Gustavo",
        "Patrícia",
    ],
)

turma_b = pd.Series(
    [9.0, 6.5, 8.5, 7.5, 9.5, 6.0, 8.0, 7.0, 9.0, 6.5, 8.5, 7.5],
    index=[
        "Ana",
        "Pedro",
        "Carla",
        "Lucas",
        "Julia",
        "Rafael",
        "Beatriz",
        "Thiago",
        "Fernanda",
        "Roberto",
        "Camila",
        "Daniel",
    ],
)


apenas_a = turma_a.index.difference(turma_b.index)
apenas_b = turma_b.index.difference(turma_a.index)
em_ambas = turma_a.index.intersection(turma_b.index)


diferenca_comum = (turma_a - turma_b).dropna()

media_turma_a = turma_a.mean()
media_turma_b = turma_b.mean()

media_comum_a = turma_a[em_ambas].mean()
media_comum_b = turma_b[em_ambas].mean()
melhor_desempenho_comum = (
    "Turma B" if media_comum_b > media_comum_a else "Turma A"
)

aluno_maior_dif_positiva = diferenca_comum.idxmax()
maior_dif_positiva = diferenca_comum.max()

aluno_maior_dif_negativa = diferenca_comum.idxmin()
maior_dif_negativa = diferenca_comum.min()


soma_direta = turma_a + turma_b
soma_com_preenchimento = turma_a.add(turma_b, fill_value=0)


print("=== a) Inspeção dos Alunos ===")
print(f"Alunos apenas na Turma A ({len(apenas_a)}): {list(apenas_a)}")
print(f"Alunos apenas na Turma B ({len(apenas_b)}): {list(apenas_b)}")
print(f"Alunos em ambas as turmas ({len(em_ambas)}): {list(em_ambas)}\n")

print("=== b) Diferença de Notas (Turma A - Turma B) ===")
print(diferenca_comum)
print()

print("=== c) Estatísticas Comparativas ===")
print(f"Média geral da Turma A: {media_turma_a:.2f}")
print(f"Média geral da Turma B: {media_turma_b:.2f}")
print(
    f"Média dos alunos em comum na Turma A: {media_comum_a:.2f} | Turma B: {media_comum_b:.2f}"
)
print(f"Melhor desempenho médio em comum: {melhor_desempenho_comum}")
print(
    f"Maior diferença positiva (A > B): {aluno_maior_dif_positiva} (+{maior_dif_positiva:.2f})"
)
print(
    f"Maior diferença negativa (B > A): {aluno_maior_dif_negativa} ({maior_dif_negativa:.2f})\n"
)

print("=== d) Soma Simples vs. Soma com fill_value=0 ===")
df_comparativo_soma = pd.DataFrame(
    {"Soma Direta (+)": soma_direta, "Soma .add(fill_value=0)": soma_com_preenchimento}
)
print(df_comparativo_soma)
````
