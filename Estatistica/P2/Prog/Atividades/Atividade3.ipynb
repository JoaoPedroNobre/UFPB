## Questão 1
Dada uma lista de n números recebida de um usuário, calcule a média aritmética e conte quantos elementos são maiores do que ela.

Input

A entrada será a seguinte:

elementos -> string

Os elementos da lista devem ser recebidos em uma única linha, separados por um espaço em branco, e devem ser convertidos para inteiro posteriormente.

Output

A saída deverá ser a quantidade de números que são maiores que a média aritmética.

Examples

Case: 1

Input
```
1 2 1 1 3 4
```

Output
```
2
```

Case: 2

Input
```
4 5 1 8 10 12 7
```

Output
```
4
```

Case: 3

Input
```
4 4 5 5 6 6 7 1 2 3
```

Output
```
5
```


````python
numeros = [int(elemento) for elemento in input().split()]

if numeros:
    media = sum(numeros) / len(numeros)
    quantidade = sum(1 for numero in numeros if numero > media)
else:
    quantidade = 0
print(quantidade)
````

## Questão 2
Escreva um programa em Python que permita ao usuário digitar uma quantidade indeterminada de números inteiros, um por vez. Depois, o programa deverá exibir a quantidade de números digitados, a quantidade de números pares e a quantidade de números ímpares.

Input

O código deve ter duas entradas:

confirmacao -> string

numero -> inteiro

A variável de confirmacao só coletará um novo número SE o usuário digitar 's' ou 'S', indicando que ele quer digitar um novo valor.

Output

A saída deverá ser:

"Quantidade de números digitados: {qtd_total}"

"Soma dos números pares: {soma_pares}"

"Soma dos números ímpares: {soma_impares}"

Examples

Case: 1

Input

```
2
s
14
s
20
n
```

Output

```
Quantidade de números digitados: 3
Soma dos números pares: 36
Soma dos números ímpares: 0
```

Case: 2

Input
```
15
s
2
s
7
s
3
n
```

Output
```
Quantidade de números digitados: 4
Soma dos números pares: 2
Soma dos números ímpares: 25
```

Case: 3

Input
```
7
s
24
s
1
s
8
y
```

Output
```
Quantidade de números digitados: 4
Soma dos números pares: 32
Soma dos números ímpares: 8
```

````python
numeros = [int(input())]
confirmacao = input()
while confirmacao.lower() == "s":
    numeros.append(int(input()))
    confirmacao = input()

soma_pares = sum(numero for numero in numeros if numero % 2 == 0)
soma_impares = sum(numero for numero in numeros if numero % 2 != 0)
print(f"Quantidade de números digitados: {len(numeros)}")
print(f"Soma dos números pares: {soma_pares}")
print(f"Soma dos números ímpares: {soma_impares}")
````
