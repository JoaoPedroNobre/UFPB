## Questão 1
Dado um número inteiro positivo n recebido de um usuário, crie um dicionário onde as chaves vão de 1 até n, e os valores são os quadrados das chaves.

Input

A entrada será:

n -> inteiro

Output

A saída deverá ser:

"{chave}: {valor}"

Para cada um dos elementos do dicionário.

Examples

Case: 1

Input
```
6
```

Output
```
1: 1
2: 4
3: 9
4: 16
5: 25
6: 36
```

Case: 2

Input
```
8
```

Output
```
1: 1
2: 4
3: 9
4: 16
5: 25
6: 36
7: 49
8: 64
```

## Questão 3
Faça um programa que armazena números de telefone de algumas pessoas, fornecidos por um usuário, em um dicionário. O programa deve receber entradas no formato nome telefone, uma por linha, até que o usuário digite "fim" para encerrar o cadastro. Depois disso, o programa deve ler um nome e imprimir o número correspondente, caso exista no dicionário.

Input

A entrada deverá ser:

linha -> string (pode ser nome telefone ou "fim")

Após o usuário digitar "fim", o programa deverá receber:

nome -> string (nome a ser consultado)

Output

A saída deverá ser:

Caso o número exista na lista telefônica: {numero}

Caso contrário: "Essa pessoa não está na lista telefônica"

**Atenção:** a variável nome pode ser tanto maiuscula quanto minuscula e mesmo assim deve exibir o número correspondente, caso exista.

Examples

Case: 1

Input
```
Ana 1248
Fabricio 1874
Leonardo 2525
fim
ana
```

Output
```
1248
```

Case: 2

Input
```
vanessa 8745
André 8989
Carlos 5847
João 178444
pedro 6487
fim
Pedro
```

Output
```
6487
```

Case: 3

Input
```
lucas 7848
emilly 4857
francisco 3648
gabriel 7485
Matheus 147155
fim
andré
```

Output
```
Essa pessoa não está na lista telefônica
```

## Questão 4
Faça um programa que receberá de um usuário nomes de pessoas, um por linha, até que o usuário digite "fim". Ao final, agrupe e imprima os nomes de acordo com a primeira letra do nome, ignorando se está em maiúsculo ou minúsculo. Os nomes devem ser agrupados em ordem alfabética da letra inicial e exibidos na ordem de inserção.

Input

A entrada deverá ser:

nome -> string (poderá ser o nome ou "fim")

Output

A saída deverá ser, para cada letra inicial que tem dentro do dicionário:

"{letra}: {nomes}"

Examples

Case: 1

Input
```
ana
Amanda
bruno
Alex
carlos
yasmin
beto
Felipe
francisco
fim
```

Output
```
A: ana Amanda Alex
B: bruno beto
C: carlos
F: Felipe francisco
Y: yasmin
```

Case: 2

Input
```
alex
alessandro
Anderson
beto
jessica
Janaina
Gerlan
fim
```

Output
```
A: alex alessandro Anderson
B: beto
G: Gerlan
J: jessica Janaina
```
