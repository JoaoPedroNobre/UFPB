## Questão 2
Faça um programa para, a partir de 10, escrever a contagem regressiva para o lançamento de um foguete. Isto é, o programa deve imprimir: 10, 9, 8, ... , 1, 0, Fogo!

````python
i = 11
for _ in range(11):                                        
    i -= 1                                                 
    print(i)                                               
print("Fogo!") 
````

## Questão 3
Faça um programa para ler 10 números inteiros e positivos e depois exibir a somatória dos números lidos.

````python
soma = 0
for i in range(1,11):
  n = int(input("Informe um número inteiro e positivo: "))
  soma += n
print(f"A soma dos números fornecidos foi {soma}.")
````

## Questão 4
Crie um programa que conte de 1 até 5 usando o laço while e exiba cada número na tela.

````python
# Inicializa a variável de controle
contador = 1

# Enquanto a condição for verdadeira...
while contador <= 5:
    print(f"Número: {contador}")
    
    # Atualiza a variável (MUITO IMPORTANTE!)
    contador = contador + 1  # ou contador += 1
````
