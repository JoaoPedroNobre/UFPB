## Questão 2

```
pessoa = {"nome": "João", "idade": 25}
```

Realize as seguintes operações:

## Questão 3
- Dado o dicionário:

```
produto = {"nome": "Notebook", "preco": 2500}
```

Utilize os métodos adequados para:

1. Tentar acessar a chave "estoque" usando get() (sem causar erro)

2. Usar setdefault() para criar a chave "estoque" com valor padrão 0

3. Exibir o dicionário após as operações

````python
produto = {"nome": "Notebook", "preco": 2500}
print(produto.get("estoque"))  # None
produto.setdefault("estoque", 0)
print(produto)  # {'nome': 'Notebook', 'preco': 2500, 'estoque': 0}
````

## Questão 4
- Dado o dicionário:

```
contato = {"nome": "Ana", "telefone": "99999-9999", "email": "ana@email.com"}
```

Peça ao usuário para digitar uma chave e verifique se ela existe no dicionário. Exiba uma mensagem informando se a chave foi encontrada ou não.

````python
contato = {"nome": "Ana", "telefone": "99999-9999", "email": "ana@email.com"}
chave = input("Digite uma chave para buscar: ")

if chave in contato.keys():
    print(f"A chave '{chave}' existe e vale: {contato[chave]}")
else:
    print(f"A chave '{chave}' não existe no dicionário")
````

## Questão 5
- Dado o dicionário:

```
notas = {"Ana": 8.5, "Bruno": 7.0, "Carla": 9.5, "Daniel": 6.0}
```

Utilizando um laço for, exiba:

1. Apenas os nomes dos alunos

2. Apenas as notas

3. O nome e a nota de cada aluno no formato: "Ana tirou nota 8.5"

````python
notas = {"Ana": 8.5, "Bruno": 7.0, "Carla": 9.5, "Daniel": 6.0}

# Apenas nomes
for nome in notas.keys():
    print(nome)

# Apenas notas
for nota in notas.values():
    print(nota)

# Nome e nota
for nome, nota in notas.items():
    print(f"{nome} tirou nota {nota}")
````

## Questão 6
- Utilizando dict comprehension, crie:

1. Um dicionário que tenha os números de 1 a 5 como chaves e seus quadrados como valores

2. Um dicionário que contenha apenas os números pares de 2 a 10 como chaves e seus dobros como valores

3. Exiba ambos os dicionários.

````python
# Quadrados de 1 a 5
quadrados = {x: x**2 for x in range(1, 6)}
print(quadrados)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Números pares de 2 a 10 e seus dobros
pares = {x: x*2 for x in range(2, 11, 2)}
print(pares)  # {2: 4, 4: 8, 6: 12, 8: 16, 10: 20}
````

## Questão 7
- Dada a lista:

```
numeros = [1, 2, 2, 3, 4, 4, 4, 5, 5, 6]
```

1. Converta a lista em um conjunto para remover as duplicatas

2. Adicione o número 7 ao conjunto

3. Remova o número 2 usando remove()

4. Exiba o conjunto final

````python
numeros = [1, 2, 2, 3, 4, 4, 4, 5, 5, 6]
conjunto = set(numeros)
print(conjunto)  # {1, 2, 3, 4, 5, 6}

conjunto.add(7)
conjunto.remove(2)
print(conjunto)  # {1, 3, 4, 5, 6, 7}
````

## Questão 8
- Dados os conjuntos:

```
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
```

Calcule e exiba:

1. A união de A e B

2. A interseção de A e B

3.  diferença A - B

4. A diferença simétrica (elementos que estão em A ou B, mas não em ambos)

````python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(A | B)  # União: {1, 2, 3, 4, 5, 6, 7, 8}
print(A & B)  # Interseção: {4, 5}
print(A - B)  # Diferença: {1, 2, 3}
print(A ^ B)  # Diferença simétrica: {1, 2, 3, 6, 7, 8}
````

## Questão 9
- Dados os conjuntos:

```
X = {2, 4, 6}
Y = {1, 2, 3, 4, 5, 6, 7}
Z = {8, 9, 10}
```

Responda (exibindo True ou False):

1. X é subconjunto de Y?

2. Y é superconjunto de X?

3. X e Z são disjuntos (não possuem elementos em comum)?

````python
X = {2, 4, 6}
Y = {1, 2, 3, 4, 5, 6, 7}
Z = {8, 9, 10}

print(X.issubset(Y))  # True
print(Y.issuperset(X))  # True
print(X.isdisjoint(Z))  # True
````

## Questão 10
- Crie um programa que:

1. Solicite ao usuário que cadastre 3 alunos

2. Para cada aluno, armazene em um dicionário: nome, idade e nota

3. Adicione cada dicionário em uma lista

4. Ao final, percorra a lista e exiba os dados de cada aluno no formato: "Nome: X, Idade: Y, Nota: Z"

````python
lista_alunos = []

for i in range(3):
    print(f"\n--- Cadastro do {i+1}º aluno ---")
    aluno = {}
    aluno["nome"] = input("Nome: ")
    aluno["idade"] = int(input("Idade: "))
    aluno["nota"] = float(input("Nota: "))
    lista_alunos.append(aluno)

print("\n--- Alunos Cadastrados ---")
for aluno in lista_alunos:
    print(f"Nome: {aluno['nome']}, Idade: {aluno['idade']}, Nota: {aluno['nota']}")
````
